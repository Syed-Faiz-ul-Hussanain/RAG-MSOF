### 2. Document Processing & Embedding Functions

Below are helper functions to load text, segment it into smaller paragraphs (chunks), and generate high-quality embeddings using the `text-embedding-004` model.

```python
import PyPDF2
import numpy as np

def load_and_chunk_document(filepath):
    """Loads a text or PDF file and splits it into logical segments."""
    if filepath.lower().endswith('.pdf'):
        text = ""
        with open(filepath, 'rb') as f:
            reader = PyPDF2.PdfReader(f)
            for page in reader.pages:
                text += page.extract_text() + "\n"
    else:
        with open(filepath, 'r', encoding='utf-8') as f:
            text = f.read()

    # Split by double newlines to keep paragraphs intact
    paragraphs = [p.strip() for p in text.split('\n\n') if p.strip()]
    if not paragraphs:
        # Fallback if no double newlines exist
        paragraphs = [p.strip() for p in text.split('\n') if p.strip()]
    return paragraphs

def get_embedding(text_to_embed):
    """Generates a vector embedding using the Gemini API."""
    response = client.models.embed_content(
        model="text-embedding-004",
        contents=text_to_embed
    )
    # Extract the embedding values
    return response.embeddings[0].values

def cosine_similarity(a, b):
    """Calculates cosine similarity between two vectors."""
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
import re
import time
import PyPDF2
import numpy as np

# Standardizing on 'gemini-embedding-001' as requested by the user
EMBED_MODEL = "gemini-embedding-001"
EMBED_DIM = 768

def load_and_chunk_document(filepath, max_chars=800, overlap=150):

    if filepath.lower().endswith(".pdf"):
        text = ""
        with open(filepath, "rb") as f:
            reader = PyPDF2.PdfReader(f)
            for page in reader.pages:
                text += (page.extract_text() or "") + "\n"
    else:
        with open(filepath, "r", encoding="utf-8") as f:
            text = f.read()

    # Normalize white spaces and segment by double/single line breaks
    paragraphs = [re.sub(r"\s+", " ", p).strip() for p in text.split("\n\n") if p.strip()]
    if len(paragraphs) <= 1:
        paragraphs = [re.sub(r"\s+", " ", p).strip() for p in text.split("\n") if p.strip()]

    pieces = []
    for p in paragraphs:
        if len(p) <= max_chars:
            pieces.append(p)
        else:
            # Split large paragraphs on sentence boundaries using positive lookbehinds
            pieces.extend(s.strip() for s in re.split(r"(?<=[.!?])\s+", p) if s.strip())

    # Re-assemble segments using a sliding-window overlap algorithm
    chunks, current = [], ""
    for piece in pieces:
        if len(current) + len(piece) + 1 <= max_chars:
            current = f"{current} {piece}".strip()
        else:
            if current:
                chunks.append(current)
            tail = current[-overlap:] if current else ""
            current = f"{tail} {piece}".strip()
    if current:
        chunks.append(current)
    return chunks

def _normalize(v):
    """
    Applies L2 normalization to vectors.

    Why vector normalization is used:
    - Cosine similarity is mathematically equivalent to the dot product of L2-normalized vectors.
    - This allows us to use fast matrix operations (@) instead of slower individual loops later.
    """
    v = np.array(v, dtype=np.float32)
    return v / np.linalg.norm(v)

def get_embeddings(texts, task="RETRIEVAL_DOCUMENT", batch_size=50):

    all_embs = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i + batch_size]
        for attempt in range(5):
            try:
                resp = client.models.embed_content(
                    model=EMBED_MODEL,
                    contents=batch,
                    config=types.EmbedContentConfig(
                        task_type=task,
                        output_dimensionality=EMBED_DIM,
                    ),
                )
                all_embs.extend(_normalize(e.values) for e in resp.embeddings)
                break
            except Exception as e:
                # Wait exponentially longer after each failure (1s, 2s, 4s, 8s, 16s)
                wait = 2 ** attempt
                print(f"Retry in {wait}s ({e.__class__.__name__})")
                time.sleep(wait)
        else:
            raise RuntimeError("Embedding failed after retries.")
    return np.array(all_embs)

def get_embedding(text, task="RETRIEVAL_QUERY"):
    """
    Utility wrapper to embed singular text blocks (e.g., incoming user search questions).
    """
    return get_embeddings([text], task=task)[0]

def cosine_similarity(a, b):
    """
    Calculates the angular difference between two multidimensional vectors.
    """
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))
```

```python
def ask_faq(query, chunks, embeddings, top_k=3):
    # 1. Obtain a 1D vector embedding representing the semantic meaning of the text query
    query_emb = get_embedding(query, task="RETRIEVAL_QUERY")

    # 2. Compute similarity scores using matrix multiplication (@ operator)
    #    This multiplies the 2D matrix of chunk embeddings (shape: N x 768) by the 1D query vector (shape: 768)
    #    resulting in a 1D array of scores (shape: N) representing the similarity of each chunk to the query
    scores = embeddings @ query_emb

    # 3. Sort indices ascendingly, reverse the order to obtain highest similarities first, and slice for top_k
    top_idx = np.argsort(scores)[::-1][:top_k]

    # 4. How this line functions:
    #    - It is a Python list comprehension that loops over each index 'i' in our filtered 'top_idx' array.
    #    - For each index 'i', it extracts the corresponding text chunk string from the 'chunks' list.
    #    - It pairs it with the score at 'scores[i]', explicitly casting it to a Python float.
    #    - It packages them together into a tuple: (chunk_text, similarity_score).
    #    - Finally, it gathers all these tuples together and returns them in a sorted list.
    return [(chunks[i], float(scores[i])) for i in top_idx]

# Define the search question targets
user_question = "who is the batch advisor for data Science ?"

try:
    results = ask_faq(user_question, chunks, chunk_embeddings)
    print(f"Question: {user_question}\n")
    for rank, (text, score) in enumerate(results, 1):
        print(f"#{rank} (Similarity: {score:.4f})\n{text}\n{'-'*60}")
except NameError:
    print("Please build the Knowledge Base in the step above before asking a question.")
```

# Project Documentation: Document FAQ Semantic Search System

This documentation outlines the architectural approach, technical pipeline implementations, and the debugging journey encountered during the development of this Question-Answering (FAQ) system.

---

## 1. Project Overview & Objective
The goal of this project is to build an in-memory document retrieval engine. The system accepts a multi-page PDF or text document, processes and segments the raw text, converts these segments into semantic vector embeddings using Google's Gemini Developer API, and performs high-speed cosine similarity searches to retrieve accurate paragraph-level answers for arbitrary user queries.

---

## 2. Technical Pipeline Approach

### Phase A: Document Loading & Robust Chunking
* **Handling Formats**: The system uses `PyPDF2` to read file streams and extract plaintext from PDF pages, falling back to basic UTF-8 string reading for standard `.txt` files.
* **Smart Text Chunking**: Raw texts are split into discrete overlapping paragraphs. Large paragraphs exceeding `max_chars=800` are split on sentence-ending punctuation (`.`, `!`, `?`) using positive lookbehinds (`(?<=[.!?])\s+`) to prevent losing grammatical boundaries.
* **Sliding Window Overlap**: To prevent loss of contextual information across chunk edges, a sliding character window overlap (`overlap=150`) is appended to consecutive chunks.

### Phase B: Batch Vector Generation
* **Embedding Model**: We utilized `gemini-embedding-001` with an output dimensionality of `768` dimensions.
* **Batch Processing**: To minimize network overhead, texts are embedded in batches (`batch_size=50`).
* **Exponential Backoff**: To handle potential rate limits (HTTP 429), an exponential wait backoff loop (`time.sleep(2 ** attempt)`) dynamically retries calls up to 5 times before failing.
* **Vector L2 Normalization**: Embedding coordinates are mathematically scaled using L2-normalization ($v / \|v\|_2$). This enforces unit length, meaning subsequent cosine similarity can be calculated utilizing highly optimized dot product operations.

### Phase C: Vector Search & Retrieval
* **Retrieval Query Context**: When a user queries the engine, the query is embedded with the `task_type="RETRIEVAL_QUERY"` parameter, while document segments use `task_type="RETRIEVAL_DOCUMENT"` for optimal cross-modal pairing.
* **Matrix Dot-Product**: Similarity scoring is computed using NumPy matrix multiplication (`embeddings @ query_emb`), which executes vectorized CPU-level operations to calculate cosine similarity instantaneously across thousands of segments.
* **List Comprehension Mapping**: The function maps the top index arguments to their original string structures and float metrics via Python list comprehensions:
  `[(chunks[i], float(scores[i])) for i in top_idx]`.
